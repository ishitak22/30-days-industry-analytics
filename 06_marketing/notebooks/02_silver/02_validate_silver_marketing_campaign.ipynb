```python
SILVER_TABLE_NAME = "workspace.marketing_campaign.silver_marketing_campaign"

silver_df = spark.table(SILVER_TABLE_NAME)

print(f"Silver table: {SILVER_TABLE_NAME}")
print(f"Rows: {silver_df.count()}")
print(f"Columns: {len(silver_df.columns)}")
```

```python
display(silver_df.limit(10))
```

```python
silver_df.printSchema()
```

```python
from pyspark.sql.functions import col, count, when

null_counts_df = silver_df.select([
    count(when(col(column_name).isNull(), column_name)).alias(column_name)
    for column_name in silver_df.columns
])

display(null_counts_df)
```

```python
display(
    silver_df
    .groupBy("response")
    .count()
    .orderBy("response")
)
```

```python
columns_to_summarize = [
    "income",
    "customer_age",
    "customer_tenure_days",
    "total_spend",
    "total_purchases"
]

display(
    silver_df
    .select(columns_to_summarize)
    .summary()
)
```

```python
required_columns = [
    "id",
    "year_birth",
    "education",
    "marital_status",
    "income",
    "dt_customer",
    "recency",
    "customer_age",
    "customer_tenure_days",
    "has_children",
    "total_spend",
    "total_purchases",
    "accepted_previous_campaign",
    "response"
]

missing_columns = [
    column_name
    for column_name in required_columns
    if column_name not in silver_df.columns
]

if missing_columns:
    print("Missing columns:")
    print(missing_columns)
else:
    print("All required columns are present.")
```

```python
invalid_age_count = silver_df.filter(
    (col("customer_age") < 18) | (col("customer_age") > 100)
).count()

invalid_income_count = silver_df.filter(col("income") <= 0).count()

invalid_response_count = silver_df.filter(~col("response").isin(0, 1)).count()

print(f"Invalid age rows: {invalid_age_count}")
print(f"Invalid income rows: {invalid_income_count}")
print(f"Invalid response rows: {invalid_response_count}")
```
