```python
SILVER_TABLE_NAME = "workspace.marketing_campaign.silver_marketing_campaign"
GOLD_TABLE_NAME = "workspace.marketing_campaign.gold_customer_features"

silver_df = spark.table(SILVER_TABLE_NAME)

print(f"Silver table: {SILVER_TABLE_NAME}")
print(f"Gold table: {GOLD_TABLE_NAME}")
print(f"Silver rows: {silver_df.count()}")
```

```python
from pyspark.sql.functions import col

gold_df = silver_df.select(
    col("id").alias("customer_id"),
    col("education"),
    col("marital_status"),
    col("income"),
    col("customer_age"),
    col("customer_tenure_days"),
    col("has_children"),
    col("recency"),
    col("total_spend"),
    col("total_purchases"),
    col("numwebvisitsmonth"),
    col("acceptedcmp1"),
    col("acceptedcmp2"),
    col("acceptedcmp3"),
    col("acceptedcmp4"),
    col("acceptedcmp5"),
    col("accepted_previous_campaign"),
    col("complain"),
    col("response")
)
```

```python
display(gold_df.limit(10))

print(f"Gold rows before saving: {gold_df.count()}")
print(f"Gold columns before saving: {len(gold_df.columns)}")
```

```python
(
    gold_df.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable(GOLD_TABLE_NAME)
)
```

```python
gold_check_df = spark.table(GOLD_TABLE_NAME)

print(f"Rows saved: {gold_check_df.count()}")
print(f"Columns saved: {len(gold_check_df.columns)}")

display(gold_check_df.limit(10))
```
