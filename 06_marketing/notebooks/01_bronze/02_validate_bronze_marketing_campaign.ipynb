```python
BRONZE_TABLE_NAME = "workspace.marketing_campaign.bronze_marketing_campaign"

bronze_df = spark.table(BRONZE_TABLE_NAME)

print(f"Bronze table: {BRONZE_TABLE_NAME}")
print(f"Rows: {bronze_df.count()}")
print(f"Columns: {len(bronze_df.columns)}")
```

```python
display(bronze_df.limit(10))
```

```python
bronze_df.printSchema()
```

```python
from pyspark.sql.functions import col, count, when

null_counts_df = bronze_df.select([
    count(when(col(column_name).isNull(), column_name)).alias(column_name)
    for column_name in bronze_df.columns
])

display(null_counts_df)
```

```python
from pyspark.sql.functions import count

duplicate_count = (
    bronze_df
    .groupBy(bronze_df.columns)
    .count()
    .filter(col("count") > 1)
    .count()
)

print(f"Duplicate row groups: {duplicate_count}")
```

```python
display(
    bronze_df
    .groupBy("Response")
    .count()
    .orderBy("Response")
)
```
