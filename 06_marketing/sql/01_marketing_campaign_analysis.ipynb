```sql
USE CATALOG workspace;
USE SCHEMA marketing_campaign;
```

```sql
SELECT
  COUNT(*) AS customer_count,
  SUM(response) AS responded_count,
  ROUND(AVG(response), 4) AS response_rate
FROM gold_customer_features;
```

```sql
SELECT
  education,
  COUNT(*) AS customer_count,
  SUM(response) AS responded_count,
  ROUND(AVG(response), 4) AS response_rate,
  ROUND(AVG(income), 2) AS avg_income,
  ROUND(AVG(total_spend), 2) AS avg_total_spend
FROM gold_customer_features
GROUP BY education
ORDER BY response_rate DESC;
```

```sql
SELECT
  marital_status,
  COUNT(*) AS customer_count,
  SUM(response) AS responded_count,
  ROUND(AVG(response), 4) AS response_rate,
  ROUND(AVG(total_spend), 2) AS avg_total_spend
FROM gold_customer_features
GROUP BY marital_status
ORDER BY response_rate DESC;
```

```sql
SELECT
  has_children,
  COUNT(*) AS customer_count,
  SUM(response) AS responded_count,
  ROUND(AVG(response), 4) AS response_rate,
  ROUND(AVG(total_spend), 2) AS avg_total_spend,
  ROUND(AVG(total_purchases), 2) AS avg_total_purchases
FROM gold_customer_features
GROUP BY has_children
ORDER BY has_children;
```

```sql
SELECT
  accepted_previous_campaign,
  COUNT(*) AS customer_count,
  SUM(response) AS responded_count,
  ROUND(AVG(response), 4) AS response_rate,
  ROUND(AVG(total_spend), 2) AS avg_total_spend
FROM gold_customer_features
GROUP BY accepted_previous_campaign
ORDER BY accepted_previous_campaign;
```

```sql
SELECT
  CASE
    WHEN income < 30000 THEN 'Under 30k'
    WHEN income < 60000 THEN '30k-59k'
    WHEN income < 90000 THEN '60k-89k'
    ELSE '90k+'
  END AS income_band,
  COUNT(*) AS customer_count,
  SUM(response) AS responded_count,
  ROUND(AVG(response), 4) AS response_rate,
  ROUND(AVG(total_spend), 2) AS avg_total_spend
FROM gold_customer_features
GROUP BY
  CASE
    WHEN income < 30000 THEN 'Under 30k'
    WHEN income < 60000 THEN '30k-59k'
    WHEN income < 90000 THEN '60k-89k'
    ELSE '90k+'
  END
ORDER BY response_rate DESC;
```

```sql
SELECT
  CASE
    WHEN customer_age < 30 THEN 'Under 30'
    WHEN customer_age < 45 THEN '30-44'
    WHEN customer_age < 60 THEN '45-59'
    ELSE '60+'
  END AS age_band,
  COUNT(*) AS customer_count,
  SUM(response) AS responded_count,
  ROUND(AVG(response), 4) AS response_rate,
  ROUND(AVG(total_spend), 2) AS avg_total_spend
FROM gold_customer_features
GROUP BY
  CASE
    WHEN customer_age < 30 THEN 'Under 30'
    WHEN customer_age < 45 THEN '30-44'
    WHEN customer_age < 60 THEN '45-59'
    ELSE '60+'
  END
ORDER BY response_rate DESC;
```

```sql
SELECT
  customer_id,
  education,
  marital_status,
  income,
  customer_age,
  total_spend,
  total_purchases,
  accepted_previous_campaign,
  response
FROM gold_customer_features
ORDER BY total_spend DESC
LIMIT 20;
```
